```python
import random

thrownDiceNumbers = []

for i in range(5):
    numero = random.randint(1, 6)
    thrownDiceNumbers.append(numero)

print("Dice numbers are: ", thrownDiceNumbers)

summattu = sum(thrownDiceNumbers)
print("Sum of the numbers is: ", summattu)

highestValue = max(thrownDiceNumbers)
print("Highest number is: ", highestValue)
```
