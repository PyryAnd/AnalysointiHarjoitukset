```python
name = "Petteri"
guessTotal = 0

guess = input("Please, guess my name.")


while guess != name:
    guess = input("Please, guess my name.")
    guessTotal += 1

    
    if guess == name:
        print ("Congratulations!")
        print("Guesses: ",guessTotal)
    else:
        quitGame = input("Do you want to quit (y/n) ?")

        if quitGame == "y":
            print("Pelit on pelattu")
            print("Guesses: ", guessTotal)
            break
```
